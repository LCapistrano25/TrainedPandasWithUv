### DataFrame Brasileiro

```python
df_soccer = pd.read_html(
    "https://gist.github.com/rogernogueira/6f6f7f165e29dc45337c4b55549c61ea"
)
```

```python
data = df_soccer[0][1].str.split(";")
columns = data[0]
```

```python
df_soccer = pd.DataFrame(data[1:].tolist(), columns=columns)
df_soccer.head()
```

```python
columns
```

```python
df_soccer.drop(columns=[''], inplace=True)
df_soccer
```

```python
df_soccer.dropna(inplace=True)
df_soccer
```

```python
df_soccer['DESCR_UNIDADE'] = df_soccer['DESCR_UNIDADE'].str.strip()
df_soccer['DESCR_UNIDADE'] = df_soccer['DESCR_UNIDADE'].apply(lambda x: np.nan if x == 'NULL' else x)
df_soccer['DESCR_UNIDADE'].unique()
```
